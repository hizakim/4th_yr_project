```python
import jax
import jax.numpy as jnp
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import pandas as pd

# Default mitral count and OSN input for symmetric params
m = 2
x = jnp.ones(m)
```

```python
# Naka-Rushton function

def nr(x, n):
    """Naka-Rushton activation function"""
    x = jnp.maximum(x, 0.0)
    return jnp.power(x, n) / (jnp.power(x, n) + 1)
```

## 1. Fixing Hill exponents from Elisa

Fit n and scale per cell class from Elisa's firing rate curves `n_r, n_b, n_g, n_y` 
(firing rates vs injected current)

Fit data to Naka Rushton

# Mitral cells

```python
df_mitral_unperturbed = pd.read_excel(
    'elisa_nr_data/Galliano et al 2021 - fig 4 ephys.xlsx',
    sheet_name='repetitive firing',
    index_col=0,
    skiprows=3,
    usecols='A,G:AB',
    nrows=45
)
df_mitral_unperturbed
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
df_mitral_unperturbed.plot(ax=ax, legend=True)
ax.set_xlabel('Injected current (pA)')
ax.set_ylabel('Firing rate (Hz)')
plt.tight_layout()
plt.show()
```

```python
df_plot = df_mitral_unperturbed
x = np.asarray(df_plot.index, dtype=float)
y = np.asarray(df_plot.mean(axis=1), dtype=float)

def nr_full(x, n, b, y_max):
    x = x / b
    return nr(x, n) * y_max

fit_params, _ = curve_fit(nr_full, x, y, p0=[3, 200, 100])
print("n = ", fit_params[0])
print("b = ", fit_params[1])
ax = df_plot.mean(axis=1).plot(figsize=(8, 6))
ax.set_xlabel('Injected current (pA)')
ax.set_ylabel('Mean firing rate (Hz)')

# plot fit
x_fit = np.linspace(x.min(), 2*x.max(), 100)
y_fit = nr_full(x_fit, *fit_params)
ax.plot(x_fit, y_fit, 'r-', label='Naka-Rushton fit')
ax.legend()
plt.tight_layout()
plt.show()
```

# Red cells (ETCs)

```python
df_red_unperturbed = pd.read_excel(
    'elisa_nr_data/Galliano et al 2021 - fig 5 ephys.xlsx',
    sheet_name='ETC repetitive firing',
    index_col=0,
    skiprows=4,
    usecols='J,N:AQ',
    nrows=35
).dropna(how='all').astype(float)
df_red_unperturbed
```

```python
df_plot = df_red_unperturbed.iloc[:-5]
x = np.asarray(df_plot.index, dtype=float)
y = np.asarray(df_plot.mean(axis=1), dtype=float)

def nr_full(x, n, b):
    x = x / b
    return nr(x, n) * y.max()

fit_params, _ = curve_fit(nr_full, x, y, p0=[3, 200])
print("n = ", fit_params[0])
print("b = ", fit_params[1])
ax = df_plot.mean(axis=1).plot(figsize=(10, 5))
ax.set_xlabel('Injected current (pA)')
ax.set_ylabel('Mean firing rate (Hz)')

# plot fit
x_fit = np.linspace(x.min(), 2*x.max(), 100)
y_fit = nr_full(x_fit, *fit_params)
ax.plot(x_fit, y_fit, 'r-', label='Naka-Rushton fit')
ax.legend()
plt.tight_layout()
plt.show()
```

# Green cells

```python
df_green_unperturbed = pd.read_excel(
    'elisa_nr_data/Galliano et al 2021 - DA cells ephys repetitive firing.xlsx',
    sheet_name='DA cells (Fig 7 and 8)',
    index_col=0,
    skiprows=4,
    usecols='A,C:L',
    nrows=51
).dropna(how='all').astype(float)
df_green_unperturbed
```

```python
df_plot = df_green_unperturbed.drop(columns=['B.4', 'D.1'])
x = np.asarray(df_plot.index, dtype=float)
y = np.asarray(df_plot.mean(axis=1), dtype=float)

def nr_full(x, n, b):
    x = x / b
    return nr(x, n) * y.max() * 1.3

fit_params, _ = curve_fit(nr_full, x, y, p0=[3, 200])

print("n = ", fit_params[0])
print("b = ", fit_params[1])
ax = df_plot.mean(axis=1).plot(figsize=(10, 5))
ax.set_xlabel('Injected current (pA)')
ax.set_ylabel('Mean firing rate (Hz)')
# plot fit
x_fit = np.linspace(x.min(), 2*x.max(), 100)
y_fit = nr_full(x_fit, *fit_params)
ax.plot(x_fit, y_fit, 'r-', label='Naka-Rushton fit')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

dfs = [
    df_red_unperturbed.iloc[:-5],
    df_green_unperturbed.drop(columns=['B.4', 'D.1']),
    df_blue_unperturbed,
    df_mitral_unperturbed
]

titles = ['Red Neurons', 'Green Neurons', 'Blue Neurons', 'Mitral Cells']
axes_list = [axes[0, 0], axes[0, 1], axes[1, 0], axes[1, 1]]

for i in range(4):
    df_plot = dfs[i]
    title = titles[i]
    ax = axes_list[i]

    x = np.asarray(df_plot.index, dtype=float)
    y = np.asarray(df_plot.mean(axis=1), dtype=float)

    # Mitral fit includes y_max parameter
   
    def nr_full(x, n, b, y_max):
        return nr(x / b, n) * y_max

    fit_params, _ = curve_fit(nr_full, x, y, p0=[3, 200, 100], maxfev=8000)

    print(f"\n{title}")
    print(f"n = {fit_params[0]:.3f}")
    print(f"b = {fit_params[1]:.3f}")
    print(f"y_max = {fit_params[2]:.3f}")

    x_fit = np.linspace(x.min(), x.max(), 300)
    y_fit = nr_full(x_fit, *fit_params)

    # Plot mean data
    ax.plot(x, y, color='blue', linewidth=2.5, label='Mean data')

    # Plot fitted curve
    ax.plot(x_fit, y_fit, color='red', linewidth=3, label='Naka-Rushton fit')
    ax.set_title(title, fontsize=20, fontweight='bold')

    # Axis limits
    ax.set_xlim(0, x.max())
    ax.set_ylim(0, max(y.max(), y_fit.max()) * 1.05)
    ax.tick_params(axis='both', labelsize=18)

    # Legend only on first subplot
    if i == 0:
        ax.legend(fontsize=18)

# Remove top-row x labels
axes[0, 0].set_xlabel('')
axes[0, 1].set_xlabel('')

# Remove right-column y labels
axes[0, 1].set_ylabel('')
axes[1, 1].set_ylabel('')

# Left-column y labels
axes[0, 0].set_ylabel('Mean firing rate (Hz)', fontsize=20)
axes[1, 0].set_ylabel('Mean firing rate (Hz)', fontsize=20)

# Bottom-row x labels
axes[1, 0].set_xlabel('Injected current (pA)', fontsize=20)
axes[1, 1].set_xlabel('Injected current (pA)', fontsize=20)

plt.tight_layout()
plt.savefig(f'Figures_final/n_fit_all_cells.eps')
plt.show()
```
